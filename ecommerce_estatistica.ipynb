## Conclusão

A análise exploratória do conjunto de dados de e-commerce permitiu identificar diferentes características dos produtos, preços, avaliações, marcas, gêneros e faixas de vendas.

A distribuição dos preços apresentou maior concentração nas faixas de aproximadamente R$ 60 a R$ 90 e R$ 130 a R$ 180, além de uma pequena concentração de produtos com preços superiores a R$ 300.

Na análise das marcas, a Lupo apresentou a maior quantidade de produtos entre as dez principais marcas identificadas no conjunto de dados.

Em relação ao gênero, os produtos classificados como Masculino e Feminino representam a maior parte do catálogo analisado, correspondendo juntos a 82,3% dos registros.

O principal destaque da análise foi a forte correlação positiva entre o número de avaliações e a faixa de vendas, com coeficiente de correlação de 0,91. O gráfico de regressão também apresentou uma tendência positiva, reforçando essa relação.

Por outro lado, variáveis como preço, desconto e nota apresentaram correlações muito mais fracas com a faixa de vendas. Dessa forma, dentro deste conjunto de dados, o número de avaliações apresentou a associação mais relevante com as faixas de vendas dos produtos.

É importante destacar que correlação não implica necessariamente causalidade. Portanto, os resultados indicam associações entre as variáveis, mas não permitem afirmar que o número de avaliações seja, isoladamente, responsável por maiores níveis de vendas.
